```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

print(X)
print(y)
```

```python
plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Training Data")
plt.show()
```

```python
m = 0
b = 0

learning_rate = 0.01
epochs = 1000
```

```python
def compute_cost(X, y, m, b):
    predictions = m * X + b
    cost = np.mean((y - predictions) ** 2)
    return cost
```

```python
cost_history = []

n = len(X)

for epoch in range(epochs):

    predictions = m * X + b

    dm = (-2/n) * np.sum(X * (y - predictions))
    db = (-2/n) * np.sum(y - predictions)

    m = m - learning_rate * dm
    b = b - learning_rate * db

    cost = compute_cost(X, y, m, b)
    cost_history.append(cost)
```

```python
print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
predictions = m * X + b

plt.scatter(X, y, label="Actual Data")
plt.plot(X, predictions, label="Regression Line")

plt.legend()
plt.show()
```

```python
plt.plot(cost_history)

plt.xlabel("Epochs")
plt.ylabel("Cost")

plt.title("Cost vs Epochs")

plt.show()
```

```python
import matplotlib.pyplot as plt
import os

# Create results folder if it doesn't exist
os.makedirs("../results", exist_ok=True)

plt.figure(figsize=(8,5))

plt.plot(cost_history)

plt.xlabel("Epoch")
plt.ylabel("Cost")
plt.title("Linear Regression: Cost vs Epoch")

# Save graph
plt.savefig("../results/linear_regression_cost_vs_epoch.png",
            dpi=300,
            bbox_inches='tight')

plt.show()
```
