# Final Results Table

This notebook consolidates the final performance metrics of all baseline
machine learning models evaluated on the Pima Diabetes dataset.

The generated `canonical_baseline_results.csv` serves as the official baseline
used by:
- Ablation Study
- README
- Research Paper
- Cross-model comparisons

```python
import pandas as pd

# Derived from results_log.csv rather than hand-typed, so this table can
# never silently drift out of sync with what the baseline notebooks
# actually computed.
log = pd.read_csv("../../results/results_log.csv")

name_map = {
    "Logistic Regression": "Logistic Regression",
    "DT": "Decision Tree",
    "RF": "Random Forest",
    "SVM": "Support Vector Machine",
    "Static MLP": "Multi-Layer Perceptron",
}

results = log.copy()
results["model"] = results["model"].map(name_map)

assert results["model"].notna().all(), (
    f"results_log.csv contains a model name not in name_map: "
    f"{log['model'][results['model'].isna()].tolist()}"
)

order = ["Logistic Regression", "Decision Tree", "Random Forest", "Support Vector Machine", "Multi-Layer Perceptron"]
results = results.set_index("model").loc[order].reset_index()

results = results.round({
    "accuracy": 4, "precision": 4, "recall": 4, "f1": 4, "auc_roc": 4
})

results
```

```python
results.to_csv(
    "../../results/canonical_baseline_results.csv",
    index=False
)

print("Saved successfully!")
```
