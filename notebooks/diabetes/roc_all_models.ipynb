```python
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.metrics import auc


lr = pd.read_csv("../../results/roc/roc_logistic.csv")
dt = pd.read_csv("../../results/roc/roc_decision_tree.csv")
rf = pd.read_csv("../../results/roc/roc_random_forest.csv")
svm = pd.read_csv("../../results/roc/roc_svm.csv")
mlp = pd.read_csv("../../results/roc/roc_mlp.csv")


auc_lr = auc(lr["FPR"], lr["TPR"])
auc_dt = auc(dt["FPR"], dt["TPR"])
auc_rf = auc(rf["FPR"], rf["TPR"])
auc_svm = auc(svm["FPR"], svm["TPR"])
auc_mlp = auc(mlp["FPR"], mlp["TPR"])


plt.figure(figsize=(8,6))

plt.plot(
    lr["FPR"],
    lr["TPR"],
    linewidth=2,
    label=f"Logistic Regression (AUC = {auc_lr:.3f})"
)

plt.plot(
    dt["FPR"],
    dt["TPR"],
    linewidth=2,
    label=f"Decision Tree (AUC = {auc_dt:.3f})"
)

plt.plot(
    rf["FPR"],
    rf["TPR"],
    linewidth=2,
    label=f"Random Forest (AUC = {auc_rf:.3f})"
)

plt.plot(
    svm["FPR"],
    svm["TPR"],
    linewidth=2,
    label=f"SVM (AUC = {auc_svm:.3f})"
)

plt.plot(
    mlp["FPR"],
    mlp["TPR"],
    linewidth=2,
    label=f"MLP (AUC = {auc_mlp:.3f})"
)

plt.plot(
    [0,1],
    [0,1],
    linestyle="--",
    color="black"
)

plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)

plt.title(
    "Receiver Operating Characteristic (ROC) Curves for All Models",
    fontsize=14
)

plt.legend(loc="lower right", fontsize=10)

plt.grid(True)


os.makedirs("../../results/roc", exist_ok=True)

plt.savefig(
    "../../results/roc/roc_all_models.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("ROC comparison figure saved successfully!")
```
